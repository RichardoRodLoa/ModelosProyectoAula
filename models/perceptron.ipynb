# Perceptrón 

```python
from libs.data_configuration import get_data_partitioned

X_train, X_test, y_train, y_test = get_data_partitioned() # 20% de datos de validacion por defecto 
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler() 

X_train_norm = scaler.fit_transform(X_train)
X_test_norm = scaler.transform(X_test)
```

```python
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score, f1_score, recall_score

alphas = [.001, .01, .1, 1, 10, 100]
penalties = ['l2','l1','elasticnet']


rows = []
for a in alphas: 
    for p in penalties: 
        
        model = Perceptron(
            alpha=a, 
            penalty=p,
            random_state=0
        )
        
        model.fit(X_train_norm, y_train)
        
        predictions_train = model.predict(X_train_norm)
        predictions_val = model.predict(X_test_norm)
        
        rows.append({
            'tiplo de penalizacion': p,
            'alpha de regularizaicion': a, 
            'acc_train': accuracy_score(y_train, predictions_train), 
            'acc_test': accuracy_score(y_test, predictions_val), 
            'f1_score': f1_score(y_test, predictions_val), 
            'recall_score': recall_score(y_test, predictions_val)
        })
```

```python
import pandas as pd 

results = pd.DataFrame(rows)
results.sort_values(by='f1_score', ascending=False)
```
