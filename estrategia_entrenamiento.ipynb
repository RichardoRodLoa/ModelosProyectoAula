# Estrategia de entrenamiento

En este notebook se explica y justifica la estrategia de entrenamiento que se utilizo, la cual es la misma para todos nuestros 5 modelos.

## Modelos Seleccionados 
- **KNN**: K vecinos mas cercanos (Modelo no parametrico)
- **Random Forest**: Modelo de ensamble de arboles
- **Perceptron Multicapa**: Modelo basado en redes neuronales
- **Regresion Logistica**: Modelo de regresion logistica (Modelo parametrico)

## Particionamiento de los datos

Para el particionamiento de los datos, usamos la librearia sklearn con la funcion `train_test_split` que nos permite particionar los datos en cuantro conjuntos: 
- `X_train`: Matriz X para entrenamiento
- `X_test`: Matriz X para pruebas
- `y_train`: vector y para entrenamiento 
- `y_test`: vector y para pruebas

Donde separamos un $20\%$ de los datos para pruebas

## Estrategia de validación 

Para la validacion de los parametros, se itera entre varias combinaciones de estos usando validacion cruzada con la clase `KFold` de la libreria sklearn, en este caso no se usa estratificación, porque como se explicó en el notebook de análisis de los datos, los datos estan perfectamente balanceados. Con una cantidad de `k=5` folds, pues es lo minimo suficiente, teniendo en cuenta los recursos computacionales que con los que contamos. 

Dentro de cada funcion de entrenamiento de los modelos, existe una lista de parametros que serán los que se iterén y se combinen con validacion crusada para encontar la mejor combinación para nuestros datos. La función de entrenamiento entrenará cada uno de estos y los organizará en una tabla, mostrando los hiperparametros que se combinarón además de los resultados de las métricas. 

## Métricas de evaluación

Como tenemos un problema de clasificación usamos 4 métricas basadas en la matríz de confusión: 

- `Accuracy` *(Para datos de entrenamiento)*: Combinados con ell accuracy de validación nos da una nocion de que tan sobre ajustados pueden llegara estar los modelos
- `Accuracy` *(Para datos de validación)*: Nos da una muestra global de el porcentaje de aciertos del modelo
- `Recall`: Nos da una nocion de los casos positivos reales que el modelo logra detectar correctamente. Queremos que el modelo detecte el mayor numero posible de páginas phishing cuando realmente lo son. 
- `F1 Score`: Nos sirve para econtar un equilibrio entre la presisión y el recall, aunque nuestra prioridad es el maximizar el recall tampoco queremos que el modelo prediga páginas legitimas como páginas de phishing
